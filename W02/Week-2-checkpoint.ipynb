Mission 6：匯入資料科學套件

```python
import numpy as np
import pandas as pd

# 設定亂數種子
random_seed = 42
rng = np.random.default_rng(random_seed)

# 測試：產生 3 個 0~1 之間的亂數
print("前 3 個亂數：", rng.random(3))

# 再執行一次：結果會一樣嗎？
print("再 3 個亂數：", rng.random(3))
```

Mission 7：建立模擬的大學生學習資料

```python
# 設定參數
n = 60  # 學生數

# 1. 學生編號 1~60
student_id = np.arange(1, n + 1)

# 2. 科系：從三個科系隨機選擇
departments = ["資訊管理", "企業管理", "金融管理"]
department = rng.choice(departments, size=n)

# 3. 睡眠時數：常態分布 平均 6.7、標準差 1.1，四捨五入到小數點後 1 位
sleep_hours = rng.normal(loc=6.7, scale=1.1, size=n).round(1)

# 4. 螢幕使用時數：常態分布 平均 5.0、標準差 1.5，限制 1~10，四捨五入到小數點後 1 位
screen_time_hours = rng.normal(loc=5.0, scale=1.5, size=n).clip(1, 10).round(1)

# 5. 讀書時數：常態分布 平均 2.5、標準差 1.0，限制 0.5~6，四捨五入到小數點後 1 位
study_hours = rng.normal(loc=2.5, scale=1.0, size=n).clip(0.5, 6).round(1)

# 6. 出席率：常態分布 平均 88、標準差 8，限制 60~100，四捨五入成整數
attendance_rate = rng.normal(loc=88, scale=8, size=n).clip(60, 100).round(0).astype(int)

# 建立 DataFrame
df = pd.DataFrame({
    "student_id": student_id,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

# 顯示前 5 筆
print("前 5 筆資料：")
print(df.head())
```

Mission 8：建立小考成績欄位

```python
# 取得各欄位方便撰寫公式
study = df["study_hours"]
sleep = df["sleep_hours"]
screen = df["screen_time_hours"]
attend = df["attendance_rate"]

# 隨機誤差
noise = rng.normal(loc=0, scale=5, size=n)

# 分項計算
base = 50
study_effect = 3 * study
sleep_effect = 2 * sleep
screen_effect = -1.2 * screen
attend_effect = 0.15 * attend

# 合計並處理
quiz_score = (base + study_effect + sleep_effect + screen_effect + attend_effect + noise).clip(0, 100).round(0).astype(int)

# 新增欄位到 DataFrame
df["quiz_score"] = quiz_score

# 顯示關鍵欄位前 5 筆
print(df[
    [
        "student_id",
        "study_hours",
        "sleep_hours",
        "screen_time_hours",
        "attendance_rate",
        "quiz_score"
    ]
].head())
```

Mission 9：加入缺失值與異常值

```python
# 故意製造缺失值：索引 5、17 的睡眠時數
df.loc[[5, 17], "sleep_hours"] = np.nan

# 故意製造異常值：索引 8 的螢幕時間設為 15.0
df.loc[8, "screen_time_hours"] = 15.0

# 找出這三筆資料確認
print("索引 5（學生 6）：")
print(df.loc[5, ["student_id", "sleep_hours", "screen_time_hours"]])

print("\n索引 17（學生 18）：")
print(df.loc[17, ["student_id", "sleep_hours", "screen_time_hours"]])

print("\n索引 8（學生 9）：")
print(df.loc[8, ["student_id", "sleep_hours", "screen_time_hours"]])
```

最終整合任務：完整可重現腳本

```python
# ==========================================
# 完整整合腳本：模擬大學生學習資料建立與檢查
# ==========================================

# 1. 匯入套件
import numpy as np
import pandas as pd

# 2. 設定亂數種子，確保可重現
random_seed = 42
rng = np.random.default_rng(random_seed)

# 3. 參數設定
n = 60  # 學生數

# 4. 建立基本欄位
student_id = np.arange(1, n + 1)

departments = ["資訊管理", "企業管理", "金融管理"]
department = rng.choice(departments, size=n)

sleep_hours = rng.normal(loc=6.7, scale=1.1, size=n).round(1)
screen_time_hours = rng.normal(loc=5.0, scale=1.5, size=n).clip(1, 10).round(1)
study_hours = rng.normal(loc=2.5, scale=1.0, size=n).clip(0.5, 6).round(1)
attendance_rate = rng.normal(loc=88, scale=8, size=n).clip(60, 100).round(0).astype(int)

# 5. 建立 DataFrame
df = pd.DataFrame({
    "student_id": student_id,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

# 6. 建立小考成績欄位
study = df["study_hours"]
sleep = df["sleep_hours"]
screen = df["screen_time_hours"]
attend = df["attendance_rate"]
noise = rng.normal(loc=0, scale=5, size=n)

quiz_score = (
    50
    + 3 * study
    + 2 * sleep
    - 1.2 * screen
    + 0.15 * attend
    + noise
).clip(0, 100).round(0).astype(int)

df["quiz_score"] = quiz_score

# 7. 加入缺失值與異常值
df.loc[[5, 17], "sleep_hours"] = np.nan      # 索引 5、17 → 學生 6、18
df.loc[8, "screen_time_hours"] = 15.0        # 索引 8 → 學生 9

# 8. 顯示前 5 筆
print("=== 前 5 筆資料 ===")
print(df.head())

# 9. 資料品質檢查
print("\n=== 形狀 ===")
print(df.shape)

print("\n=== 欄位 ===")
print(df.columns)

print("\n=== 型態 ===")
print(df.dtypes)

print("\n=== 缺失值統計 ===")
print(df.isna().sum())

print("\n=== 描述統計 ===")
print(df.describe())

# 10. 匯出 CSV
file_name = "student_learning_data.csv"
df.to_csv(file_name, index=False, encoding="utf-8-sig")
print(f"\n資料已儲存為：{file_name}")

# 若在 Colab，取消註解下兩行以下載
# from google.colab import files
# files.download(file_name)
```
