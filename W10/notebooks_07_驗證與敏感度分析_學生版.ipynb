以下在記憶體重建與既有教材相同的資料。執行後可以直接開始本單元；不會覆寫 CSV。


````python
# 匯入套件：numpy 建立隨機資料，pandas 管理表格，pyplot 繪圖
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 固定種子，並維持呼叫順序，以重現原教材的 60 筆模擬資料
rng = np.random.default_rng(42)
n = 60
departments = ["資訊管理", "企業管理", "金融管理"]
# 每個字典鍵是一個欄位；對應的值是 60 名學生的資料
data = {
    "student_id": range(1, n + 1),
    "department": rng.choice(departments, n),
    "sleep_hours": rng.normal(6.7, 1.1, n).round(1),
    "screen_time_hours": rng.normal(5.0, 1.5, n).clip(1, 10).round(1),
    "study_hours": rng.normal(2.5, 1.0, n).clip(0.5, 6).round(1),
    "attendance_rate": rng.normal(88, 8, n).clip(60, 100).round(0),
}
df = pd.DataFrame(data)
# 成績依設定公式與隨機誤差建立；關係來自模擬規則，不能視為因果發現
noise = rng.normal(0, 5, n)
score = (50 + 3 * df["study_hours"] + 2 * df["sleep_hours"]
         - 1.2 * df["screen_time_hours"] + 0.15 * df["attendance_rate"] + noise)
df["quiz_score"] = score.clip(0, 100).round(0)
# 故意在計分後加入資料品質問題；無法從表格還原計分時原始值
df.loc[[5, 17], "sleep_hours"] = np.nan
df.loc[8, "screen_time_hours"] = 15.0
print("模擬資料形狀（列, 欄）：", df.shape)
````

### AI 回答重點（請自行整理）：
**敏感度分析（Sensitivity Analysis）可以理解為資料分析中的「壓力測試」**
 1. 步驟一：敏感度分析
 - 敏感度分析，同時比較方案 A（保留所有有效資料）與方案 B（排除極端值）的結果，結果通常會落在以下兩種情況之一：
 - 敏感度低：不管有沒有刪掉極端值，算出來的相關係數都很接近（例如從 $-0.65$ 變成 $-0.62$），結論方向完全一致。代表結論很可靠，不會被少數幾筆特殊資料左右。
 - 敏感度高：只要刪掉一兩筆極端值，相關係數就大幅翻轉（例如從 $-0.65$ 變成 $-0.05$）。這是在提醒你：結論高度依賴那幾筆極端資料，下結論時須謹慎。

 2. 步驟二：為什麼「改變極端值處理」不自動證明何者正確？
 - 情況一：極端值是「人為輸入錯誤或系統異常」（應該排除）
 - 例子：問卷調查「每天使用手機螢幕幾小時」，有一筆填了 99 小時。這時排除大於 $24$ 小時的資料是合理的，因為它不是真實行為。
 - 情況二：極端值是「真實發生的特殊案例」（不應隨便刪除）
 - 例子：如果填 14 小時的人是一位重度手遊玩家或工作需要整天盯著螢幕的工程師，這就是真實存在的高使用量族群。如果只是為了讓圖表好看或讓相關係數變高而把大於 $10$ 小時的資料刪掉，反而會掩蓋真實世界的多樣性。

### 我的理解與具體例子：
1. **敏感度分析的核心目的（結論穩健度測試）**：
* 透過比較不同資料處理規則（如「保留」與「排除」極端值）的結果，檢驗結論是否穩健。若差異大，代表結論高度依賴少數特殊樣本，解讀時須格外謹慎。

2. **為何改變極端值處理不自動證明何者正確**：
* 敏感度分析只能揭露數據波動，無法自動判定對錯；是否排除極端值，必須回歸**資料真實性與領域邏輯**來判斷。

  - 情境：在第 3 個儲存格算好了 總分，在第 4 個儲存格修改了 期中考 加分，然後直接跳到第 5 個儲存格畫圖，沒有回去重跑第 3 個儲存格。

  - 分析失真：畫出來的圖表或相關係數使用的是「舊的總分」搭配「新的期中考分數」，導致前後分析無法對齊，別人也無法重現的結果。
 
**例子**

import pandas as pd

#1. 建立初始成績資料

df = pd.DataFrame({

    "學生": ["小王", "小李", "小陳"],

    "期末考": [80, 60, 95]
    
})

#2. 當時進行計分：計算「總分」欄位（此時總分反映的是初始分數）

df["總分"] = df["期中考"] + df["期末考"]

print(df)

df.loc[df["學生"] == "小李", "期末考"] = 85

#注意：這裡刻意「沒有」重新計算 df["總分"]

print("\n=== 2. 修改小李期末考後（未重算總分） ===")

print(df)

#4. 自行驗證：檢查「目前欄位相加」與「當時記錄的總分」是否一致

recalculated_total = df["期中考"] + df["期末考"]

### 我的理解與具體例子：
1. **靜態快照與欄位不同步的成因**：
* 在 Python 的 `pandas` 套件中，透過運算產生的衍生欄位（如總分、平均分數）只會保留執行該行程式碼當下的「靜態快照」，不會像 Excel 公式一樣在原始欄位變動後自動連動更新。若在計分後才修正原始欄位的錯誤或補齊缺漏值，卻未重新執行計分程式碼，就會導致「目前原始欄位」與「當時計算出的結果欄位」產生前後矛盾，進而破壞分析的可信度與可重現性。
* **具體例子**：假設小李的期中考為 70 分、期末考初始登錄為 60 分，執行程式算出「總分」欄位為 130 分。事後發現小李的期末考登錄錯誤，將該欄位更正為 85 分，但沒有重新執行總分計算的程式碼；此時資料表上就會出現期中 70 分、期末 85 分，總分卻仍停留在舊數值 130 分的不一致狀態。

2. **預防與檢核機制**：
* 為避免欄位不一致，資料處理應嚴格遵守「先清理、修正與補值，最後才計算衍生欄位」的順序；若中途修改了任何原始數值，必須重新由上而下執行完整流程，或透過重新計算並比對布林值（如 `df["總分"] == (df["期中考"] + df["期末考"])`）來自動抓出不同步的異常資料列。
