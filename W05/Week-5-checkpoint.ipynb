## Mission 8：建立小考成績欄位

**Knowledge Point**：欄位運算、隨機誤差、多欄選取、分數限制

沿用 Mission 4 的人為公式，另外加入平均數 0、標準差 5 的隨機誤差。`df["欄名"]` 選一欄；`df[["欄1", "欄2"]]` 選多欄。

**執行前預測**：新欄位加入後，資料表共有幾欄？所有分數應落在哪個範圍？

````python
# %% Cell 1：匯入套件
import numpy as np
import pandas as pd
from IPython.display import display

# %% Cell 2：設定亂數種子與人數
rng = np.random.default_rng(42)
n = 200

# %% Cell 3：建立資料表 df（Mission 4）
df = pd.DataFrame({
    "student_id": [f"S{i:03d}" for i in range(1, n + 1)],
    "study_hours": rng.uniform(0, 10, size=n).round(1),
    "sleep_hours": rng.uniform(4, 10, size=n).round(1),
    "screen_time_hours": rng.uniform(0, 8, size=n).round(1),
    "attendance_rate": rng.uniform(60, 100, size=n).round(1),
})
display(df.head())
print(df.shape)  # 預期 (200, 5)

# %% Cell 5：沿用 Mission 4 的人為公式，逐項計算
study_part = 3 * df["study_hours"]
sleep_part = 2 * df["sleep_hours"]
screen_part = -1.2 * df["screen_time_hours"]
attendance_part = 0.15 * df["attendance_rate"]
random_error = rng.normal(0, 5, size=n)

assert sleep_part is not None and screen_part is not None, "請先補完欄位運算。"
raw_scores = 50 + study_part
raw_scores = raw_scores + sleep_part
raw_scores = raw_scores + screen_part
raw_scores = raw_scores + attendance_part
raw_scores = raw_scores + random_error

# 限制在 0 到 100，四捨五入，再轉成整數
df["quiz_score"] = raw_scores.clip(0, 100).round().astype(int)
assert df["quiz_score"].notna().all(), "請先建立 quiz_score。"
assert df["quiz_score"].between(0, 100).all(), "分數應在 0 到 100。"
display(df[["student_id", "study_hours", "quiz_score"]].head(60))
````
