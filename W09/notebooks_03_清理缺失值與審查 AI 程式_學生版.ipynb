````python
# 匯入套件：numpy 建立隨機資料，pandas 管理表格，pyplot 繪圖
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 固定種子，並維持呼叫順序，以重現原教材的 60 筆模擬資料
rng = np.random.default_rng(42)
n = 60
departments = ["資訊管理", "企業管理", "金融管理"]
# 每個字典鍵是一個欄位；對應的值是 60 名學生的資料
data = {
    "student_id": range(1, n + 1),
    "department": rng.choice(departments, n),
    "sleep_hours": rng.normal(6.7, 1.1, n).round(1),
    "screen_time_hours": rng.normal(5.0, 1.5, n).clip(1, 10).round(1),
    "study_hours": rng.normal(2.5, 1.0, n).clip(0.5, 6).round(1),
    "attendance_rate": rng.normal(88, 8, n).clip(60, 100).round(0),
}
df = pd.DataFrame(data)
# 成績依設定公式與隨機誤差建立；關係來自模擬規則，不能視為因果發現
noise = rng.normal(0, 5, n)
score = (50 + 3 * df["study_hours"] + 2 * df["sleep_hours"]
         - 1.2 * df["screen_time_hours"] + 0.15 * df["attendance_rate"] + noise)
df["quiz_score"] = score.clip(0, 100).round(0)
# 故意在計分後加入資料品質問題；無法從表格還原計分時原始值
df.loc[[5, 17], "sleep_hours"] = np.nan
df.loc[8, "screen_time_hours"] = 15.0
print("模擬資料形狀（列, 欄）：", df.shape)
````

- **一、 查證來源與原理說明**
    - 1. 排除缺失值（Drop NA / Complete-case Analysis）的統計限制
        - 來源依據：
        - Donald B. Rubin (1976) Inference and Missing Data（缺失機制理論：MCAR 完全隨機缺失、MAR 隨機缺失、MNAR 非隨機缺失）。
        - pandas 官方文件：pandas.Series.dropna API Reference
        - 說明方式： 只有在資料符合「完全隨機缺失（MCAR）」時，直接刪除缺失列才不會導致估算結果產生系統性偏差（Selection Bias）。若缺失與特定行為有關（如熬夜者傾向不填寫），排除資料將扭曲母體分配並導致統計檢定力（Statistical Power）下降。
    - 2. 中位數填補（Median Imputation）的特性與限制
        - 來源依據：
        - Little, R. J., & Rubin, D. B. (2019). Statistical Analysis with Missing Data.
        - pandas 官方文件：pandas.Series.fillna API Reference 與 pandas.Series.std
        - 說明方式： 單一值填補（Single Imputation，包含平均數與中位數填補）雖能保留樣本數並降低離群值影響，但在數學上會將新樣本點集中於分布中心，導致樣本變異數（Variance）與標準差（Standard Deviation）人為縮小（Underestimation of Variance），同時弱化該變數與其他變數之間的共變異數及相關性。
